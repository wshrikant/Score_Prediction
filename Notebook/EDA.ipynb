```python
df['Total score']=df['math_score']+df['reading_score']+df['writing_score']
df['average score']=df['Total score']/3
df.head()
```

```python
from sklearn.metrics import mean_squared_error,r2_score,mean_absolute_error
from sklearn.neighbors import KNeighborsRegressor
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor,AdaBoostRegressor
from sklearn.svm import SVR
from sklearn.linear_model import LinearRegression,Ridge,Lasso
from sklearn.model_selection import RandomizedSearchCV
from catboost import CatBoostRegressor
from xgboost import XGBRegressor
import  warnings
```

```python
df.head()
```

```python
new_df=df.drop(['math_score','reading_score','writing_score','average score'],axis=1)
```

```python
X=new_df.drop(['Total score'],axis=1)
Y=new_df['Total score']
```

```python
X.head()
```

```python
num_feat=X.select_dtypes(exclude="object").columns
cat_feat=X.select_dtypes(include="object").columns
print(num_feat,"----",cat_feat)
```

```python
#all are cat variable in independent variable 
from sklearn.preprocessing import OneHotEncoder
```

```python
cat_encoder=OneHotEncoder(sparse_output=False)
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(X,Y,test_size=.30,random_state=42)
```

```python
X_train=cat_encoder.fit_transform(x_train)
X_test=cat_encoder.transform(x_test)
```

```python
def evaluate_model(true,predict):
    mae=mean_absolute_error(true,predict)
    mse=mean_squared_error(true,predict)
    score=r2_score(true,predict)
    return mae,mse,score
```

```python
models={
    "Linear Regression": LinearRegression(),
    "Lasso":Lasso(),
    "Ridge":Ridge(),
    "decision tree":DecisionTreeRegressor(),
    "XGB Regre":XGBRegressor(),
    "catboost":CatBoostRegressor(),
    "adaboost":AdaBoostRegressor()
}

model_list=[]
score=[]
for i in range(len(list(models))):
    model=list(models.values())[i]
    model.fit(X_train,y_train)  

    y_train_pred=model.predict(X_train)
    y_test_pred=model.predict(X_test)

    train_mae,train_mse,train_score=evaluate_model(y_train,y_train_pred)
    test_mae,test_mse,test_score=evaluate_model(y_test,y_test_pred)
    

    print(list(models.keys())[i])
    model_list.append(list(models.keys())[i])
    print("training accuracy")
    print("train_mae",train_mae)
    print("train_mse",train_mse)
    print("train_score",train_score)
    print("   ")
    print("test accuracy")
    print("test_mae",test_mae)
    print("test_mse",test_mse)
    print("test_score",test_score)
    print("--------------------")
```
